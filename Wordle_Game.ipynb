```python
from IPython.display import clear_output

import random
import colorama


WORDS = ["APPLE", "GRAPE", "PEACH", "MANGO", "BERRY", "LEMON", "CHILI", "MELON"]

secret = random.choice(WORDS)
attempts = 5

board = []

def print_hello():
    print("Вітаю у грі WORDLE!")
    print("Вгадай слово з 5 літер. У тебе 5 спроб")

print_hello()


for turn in range(1, attempts + 1):
    guess = input().upper()


    if len(guess) != len(secret):
        print("Слово має бути довжиною 5 літер")
        continue
    if not guess.isalpha():
      print("Слово має містити лише літери")
      continue




    result = ""
    for i in range(len(secret)):
        if guess[i] == secret[i]:
            result += colorama.Back.GREEN
        elif guess[i] in secret:
            result += colorama.Back.YELLOW
        else:
            result += colorama.Back.LIGHTBLACK_EX
        result += colorama.Fore.WHITE + " " + guess[i] + " " + colorama.Style.RESET_ALL


    board.append(result)

    clear_output()

    print_hello()

    for r in board:
      print(r)
    print()

    if guess == secret:
        print(f"Перемога! Слово було: {secret}")
        break
else:
    print(f" Ти не вгадав. Слово було: {secret}. Спробуй ще раз")
```
