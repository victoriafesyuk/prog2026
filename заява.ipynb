```python
import gradio as gr

def enter_date(full_name, adress, phone_number, date, receiver_name, time, belongings):

  document = f"""
Прізвище, ім’я, по батькові: {full_name}

Місце проживання (адреса): {adress}

Контактний телефон: {phone_number}

Дата: {date}

                                        ЗАЯВА

                                Про вчинення крадіжки

Шановний(а) {receiver_name} !

Прошу прийняти заяву про вчинення крадіжки.

У період {time} {date} в моєму житловому приміщенні за адресою: {adress} вчинено крадіжку майна.

Опис викраденого майна:
{belongings}

Заявник проінформований про право на захист особистих даних згідно з вимогами Закону України “Про захист персональних даних”.

Прошу вжити необхідних заходів для встановлення обставин крадіжки та притягнення винних осіб до відповідальності згідно з чинним законодавством України.

З повагою,

{full_name}

  """

  return document


demo = gr.Interface(
    fn=enter_date,
    inputs=[
        gr.Textbox(label="ПІБ"),
        gr.Textbox(label="Адреса"),
        gr.Textbox(label="Контактний телефон"),
        gr.Textbox(label="Дата"),
        gr.Textbox(label="Ім'я керівника відділу поліції"),
        gr.Textbox(label="Період крадіжки"),
        gr.Textbox(label="Опис викраденого майна")
    ],
    outputs=gr.Textbox(lines=30),
    title="Генератор заяви про крадіжку"
)

demo.launch(share=True)




```
